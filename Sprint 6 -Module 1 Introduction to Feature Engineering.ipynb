```python
## What is Feature Engineering?
### Step 1 – Understand the Concept (Markdown)
Definition; Feature Engineering is the process of creating new features, modifying existing features, or transforming features from the available data to help a Machine Learning model make better predictions.

In simple words; Feature Engineering = Creating useful information from existing data.

The original dataset may not contain all the information that is useful for a model. We can create new columns from existing columns.

Main Feature Engineering techniques
Creating new features
Combining existing features
Extracting information
Transforming features
Binning
Date/time feature extraction
Domain-based features
```

## Good Features vs Bad Features
### Step 1 – Understand the Concept (Markdown)
Definition; Good features are useful and relevant inputs that help the ML model make accurate predictions.

Bad features are irrelevant, misleading, duplicate, or problematic inputs that can reduce model performance.

In simple words

Good Feature → Helps the model.
Bad Feature → Does not help or can confuse the model.

```python
### Feature Extraction
#### Step 1 – Understand the Concept (Markdown)
Definition

Feature Extraction means extracting or creating important information from existing data and representing it as new features.

In simple words: Take useful information from existing data and create a new feature.

#### Step 2 – Demonstrate the Concept (Markdown)

Example:

Date = 2026-08-27
        ↓
Extract
        ↓
Year = 2026
Month = 8
Day = 27

Another example:

Area_sqft + Bedrooms
        ↓
Area_per_Bedroom
### Step 3 – Implement the Concept (Python)
df["Area_per_Bedroom"] = (
    df["Area_sqft"] / df["Bedrooms"]
)
```

```python
| **Topic**                        | **Definition**                                                   | **Simple Example**                               |
| -------------------------------- | ---------------------------------------------------------------- | ------------------------------------------------ |
| **Feature**                      | Input variable used for prediction                               | `Age`, `Area_sqft`                               |
| **Feature vs Variable**          | Variable = any column; Feature = variable used for ML            | `House_ID` = Variable, `Area_sqft` = Feature     |
| **Feature Engineering**          | Creating useful new features from existing data                  | `Bedrooms + Bathrooms → Total_Rooms`             |
| **Domain Knowledge**             | Understanding the real-world meaning of data                     | `Age = -5 → Invalid`                             |
| **Good vs Bad Features**         | Good features help prediction; bad features are irrelevant       | `Area_sqft` = Good, `House_ID` = Bad             |
| **Feature Engineering Workflow** | Create → Transform → Select useful features                      | `Area / Bedrooms → Area_per_Bedroom`             |
| **Feature Transformation**       | Changing the values/distribution of a feature                    | `Area → Log(Area)`                               |
| **Feature Extraction**           | Extracting useful information from existing data                 | `Date → Year, Month`                             |
| **Feature Leakage**              | Using information that should not be available during prediction | `Price → Price_per_sqft` when predicting `Price` |
```

```python
Feature          → Input for ML
Variable         → Any column
Feature Engineer → Create useful features
Domain Knowledge → Understand the data
Good Feature     → Helps prediction
Bad Feature      → Irrelevant feature
Transformation   → Change feature values
Extraction       → Extract useful information
Leakage          → Model gets future/target information
```

```python
| **Stage**                  | **Role of Feature Engineering**             | **Example**                          |
| -------------------------- | ------------------------------------------- | ------------------------------------ |
| **1. Data Collection**     | Understand available data                   | House data                           |
| **2. Data Cleaning**       | Fix missing/incorrect values                | Handle missing `Age`                 |
| **3. Feature Engineering** | Create or transform useful features         | `Bedrooms + Bathrooms → Total_Rooms` |
| **4. Feature Selection**   | Keep important features                     | Keep `Area_sqft`, remove `House_ID`  |
| **5. Encoding**            | Convert categorical data into numbers       | `Male/Female → 0/1`                  |
| **6. Scaling**             | Bring numerical features to a similar scale | `StandardScaler()`                   |
| **7. Model Training**      | Give prepared features to the model         | Train ML model                       |
| **8. Prediction**          | Use features to make predictions            | Predict house price                  |
```
