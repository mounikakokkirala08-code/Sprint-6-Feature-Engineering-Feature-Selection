### Mathematical Transformations

## Step 1 – Understand the Concept (Markdown)
Definition; Mathematical Transformation means applying a mathematical function to a feature to change its values or distribution.

In simple words: We use mathematical formulas to make data more suitable for Machine Learning.

## Step 2 – Demonstrate the Concept (Markdown)
Transformation	Formula	Use	Example
Log Transformation	log(x)	   Reduces   right skewness	Income → log(Income)
Square Root	  √x	  Reduces skewness	Income → √Income

### Step 3 – Implement the Concept (Python)

```python
import pandas as pd

df = pd.read_csv("house_price_prediction_cleaned.csv")
df.columns
```

```python
#Log Transformation


import numpy as np

df["Area_log"] = np.log1p(df["Area_sqft"])
#Square Root Transformation
df["Area_sqrt"] = np.sqrt(df["Area_sqft"])
#Square Transformation
df["Age_squared"] = df["Age_years"] ** 2
```

```python
X = df.drop(columns=['Price_lakhs', 'House_ ID'])
y = df['Price_lakhs']

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
# Price per square foot
df['Price_per_sqft'] = (df['Price_lakhs'] * 100000) / df['Area_sqft']

# Total rooms
df['Total_Rooms'] = df['Bedrooms'] + df['Bathrooms']

# Area per bedroom
df['Area_per_Bedroom'] = df['Area_sqft'] / df['Bedrooms']

# Property age category
df['Age_Category'] = pd.cut(
    df['Age_years'],
    bins=[0, 5, 15, 30, 100],
    labels=['New', 'Moderate', 'Old', 'Very Old']
)
```

### Subtraction
## Step 1 – Understand the Concept (Markdown)

Subtraction means finding the difference between two numerical features.

Formula: New Feature = Feature 1 − Feature 2

It can help create a new feature that represents the difference or gap between two variables.

## Step 2 – Demonstrate the Concept (Markdown)

For the house dataset:

Bedrooms = 3
Bathrooms = 2

Bedrooms − Bathrooms = 3 − 2 = 1

So, we can create:

Bedrooms_minus_Bathrooms = Bedrooms − Bathrooms
## Step 3 – Implement the Concept (Python)


```python
# Create a new feature using subtraction
df['Bedrooms_minus_Bathrooms'] = (
    df['Bedrooms'] - df['Bathrooms']
)

# Display the result
print(df[['Bedrooms', 'Bathrooms', 'Bedrooms_minus_Bathrooms']].head())
```

## Multiplication
### Step 1 – Understand the Concept (Markdown)

Multiplication means multiplying two numerical features to create a new feature.

Formula: New Feature = Feature 1 × Feature 2

It can help capture the combined effect of two features.

### Step 2 – Demonstrate the Concept (Markdown)

For the house dataset:

Bedrooms = 3
Bathrooms = 2

Bedrooms × Bathrooms = 3 × 2 = 6

So, we can create:

Bedroom_Bathroom_Product = Bedrooms × Bathrooms
### Step 3 – Implement the Concept (Python)


```python
# Create a new feature using multiplication
df['Bedroom_Bathroom_Product'] = (
    df['Bedrooms'] * df['Bathrooms']
)

# Display the result
print(df[['Bedrooms', 'Bathrooms', 'Bedroom_Bathroom_Product']].head())
```

## Division
### Step 1 – Understand the Concept (Markdown)

Division means dividing one numerical feature by another to create a new feature.

Formula: New Feature = Feature 1 ÷ Feature 2

It can help identify ratios and relationships between features.

### Step 2 – Demonstrate the Concept (Markdown)

For the house dataset:

Area_sqft = 1500
Bedrooms = 3

1500 ÷ 3 = 500

So:

Area_per_Bedroom = Area_sqft ÷ Bedrooms
### Step 3 – Implement the Concept (Python)


```python
# Create a new feature using division
df['Area_per_Bedroom'] = (
    df['Area_sqft'] / df['Bedrooms']
)

# Display the result
print(df[['Area_sqft', 'Bedrooms', 'Area_per_Bedroom']].head())
```

```python
| Technique                      | Formula / Example              | Purpose                         |
| ------------------------------ | ------------------------------ | ------------------------------- |
| **Percentages**                | `Bedrooms / Total_Rooms × 100` | Convert values into percentages |
| **Differences**                | `Bedrooms - Bathrooms`         | Find the gap between features   |
| **Aggregations**               | `Bedrooms + Bathrooms`         | Combine multiple features       |
| **Log Transformation**         | `log1p(Area_sqft)`             | Reduce skewness                 |
| **Square Root Transformation** | `sqrt(Area_sqft)`              | Reduce moderate skewness        |
| **Power Transformation**       | `Area_sqft ** 2`               | Capture nonlinear relationships |
| **Absolute Difference**        | `abs(Bedrooms - Bathrooms)`    | Find positive difference        |
| **Relative Difference**        | `abs(A-B) / ((A+B)/2)`         | Compare features proportionally |
| **Rate Features**              | `Area_sqft / Age_years`        | Calculate a rate or ratio       |
| **Normalized Features**        | `(X - min) / (max - min)`      | Scale values to 0–1             |
```

### Log Transformation
### Step 1 – Understand the Concept (Markdown)

Log Transformation applies a logarithmic function to numerical data.

Formula:

New Feature = log(1 + X)

It is mainly used to reduce skewness and make highly spread-out data more manageable.

### Step 2 – Demonstrate the Concept (Markdown)

For example:

Area_sqft = 1500

log(1 + 1500) ≈ 7.31

So:

Log_Area = log(1 + Area_sqft)
### Step 3 – Implement the Concept (Python)


```python
import numpy as np

df['Log_Area'] = np.log1p(df['Area_sqft'])

print(df[['Area_sqft', 'Log_Area']].head())
```

### Power Transformation
### Step 1 – Understand the Concept (Markdown)

Power Transformation applies a power to a numerical feature.

Formula:

New Feature = X^n

It is used to create nonlinear relationships and change the distribution of data.

### Step 2 – Demonstrate the Concept (Markdown)

For example:

Area_sqft = 1000

1000² = 1,000,000

So:

Power_Area = Area_sqft²
### Step 3 – Implement the Concept (Python)
df['Power_Area'] = df['Area_sqft'] ** 2

print(df[['Area_sqft', 'Power_Area']].head())

### Relative Difference
Relative Difference
### Step 1 – Understand the Concept (Markdown)

Relative Difference measures the difference between two values compared to their average.

Formula:

Relative Difference = |A - B| / ((A + B) / 2)

It shows the difference proportionally rather than as an absolute value.

### Step 2 – Demonstrate the Concept (Markdown)

For example:

Bedrooms = 4
Bathrooms = 2

|4 - 2| / ((4 + 2) / 2)
= 2 / 3
= 0.67
### Step 3 – Implement the Concept (Python)
df['Relative_Difference'] = (
    abs(df['Bedrooms'] - df['Bathrooms']) /
    ((df['Bedrooms'] + df['Bathrooms']) / 2)
)

print(df[['Bedrooms', 'Bathrooms', 'Relative_Difference']].head())

### Rate Features
### Step 1 – Understand the Concept (Markdown)

Rate Features represent how much of one feature exists relative to another.

Formula:

Rate = Feature 1 ÷ Feature 2

They are useful for capturing ratios, efficiency, or per-unit values.

### Step 2 – Demonstrate the Concept (Markdown)

For the house dataset:

Area_sqft = 1500
Bedrooms = 3

Area per Bedroom = 1500 ÷ 3 = 500

So:

Area_per_Bedroom = Area_sqft ÷ Bedrooms
### Step 3 – Implement the Concept (Python)


```python
df['Area_per_Bedroom'] = (
    df['Area_sqft'] / df['Bedrooms']
)

print(df[['Area_sqft', 'Bedrooms', 'Area_per_Bedroom']].head())
```

### Normalized Features
### Step 1 – Understand the Concept (Markdown)

Normalization scales numerical values to a common range, usually 0 to 1.

Formula:

Normalized Value = (X - Min) / (Max - Min)
### Step 2 – Demonstrate the Concept (Markdown)

Example:

Area = 1500
Min = 500
Max = 2500

(1500 - 500) / (2500 - 500)
= 0.5

So, 1500 sqft becomes 0.5.

### Step 3 – Implement the Concept (Python)
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

df['Normalized_Area'] = scaler.fit_transform(
    df[['Area_sqft']]
)

print(df[['Area_sqft', 'Normalized_Area']].head())
