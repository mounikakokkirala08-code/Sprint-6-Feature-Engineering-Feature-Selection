```python
## Categorical Variables & Encoding — With Coding
### 1. Categorical Variables
Step 1 – Understand

Categorical variables contain categories or labels instead of continuous numerical values.

Examples from your dataset:

Property_Type
Furnishing_Status
Construction_Status
Locality_Type
### Step 2 – Demonstrate
Property_Type
--------------
Apartment
Villa
Independent House
### Step 3 – Implement
```

```python
import pandas as pd
df = pd.read_csv("house_price_prediction_cleaned.csv")
```

```python
categorical_columns = [
    'Property_Type',
    'Furnishing_Status',
    'Construction_Status',
    'Locality_Type'
]

print(df[categorical_columns].head())
```

```python
## 2. Nominal Variables
### Step 1 – Understand

Nominal categories have no natural order.

Example:

Apartment, Villa, Independent House
### Step 2 – Demonstrate
Apartment → No ranking
Villa     → No ranking
House     → No ranking
### Step 3 – Implement

Use One-Hot Encoding:
```

```python

from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(handle_unknown='ignore')

encoded = encoder.fit_transform(
    df[['Property_Type']]
)

print(encoded.toarray())
```

## 3. Ordinal Variables
### Step 1 – Understand

Ordinal variables have a meaningful order.

Example:

Unfurnished < Semi-Furnished < Furnished
### Step 2 – Demonstrate
Unfurnished    → 0
Semi-Furnished → 1
Furnished      → 2
### Step 3 – Implement


```python
from sklearn.preprocessing import OrdinalEncoder

encoder = OrdinalEncoder(
    categories=[['Unfurnished', 'Semi_Furnished', 'Furnished']]
)

df['Furnishing_Encoded'] = encoder.fit_transform(
    df[['Furnishing_Status']]
)

#Use: Categories with a genuine ranking.
```

```python
4. Binary Variables
Step 1 – Understand

Binary variables have only two categories.

Example:

Yes / No
Step 2 – Demonstrate
Yes → 1
No  → 0
Step 3 – Implement
df['Parking'] = df['Parking'].map({
    'Yes': 1,
    'No': 0
})
```

```python
### 5. One-Hot Encoding
### Step 1 – Understand

Creates a separate 0/1 column for every category.

### Step 2 – Demonstrate
Property_Type

Apartment → 1 0 0
Villa     → 0 1 0
House     → 0 0 1
### Step 3 – Implement
```

```python
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(
    handle_unknown='ignore',
    sparse_output=False
)

result = encoder.fit_transform(
    df[['Property_Type']]
)

print(result)
```

### 6. Ordinal Encoding
Step 1 – Understand

Converts ordered categories into numerical values.

### Step 2 – Demonstrate
Unfurnished    → 0
Semi-Furnished → 1
Furnished      → 2
### Step 3 – Implement


```python
from sklearn.preprocessing import OrdinalEncoder

encoder = OrdinalEncoder(
    categories=[['Unfurnished', 'Semi-Furnished', 'Furnished']]
)

df['Furnishing_Encoded'] = encoder.fit_transform(
    df[['Furnishing_Status']]
)
```

### 7. Label Encoding
### Step 1 – Understand

Assigns a numerical label to each category.

### Step 2 – Demonstrate
Apartment → 0
House     → 1
Villa     → 2
### Step 3 – Implement


```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

df['Property_Type_Encoded'] = encoder.fit_transform(
    df['Property_Type']
)
```

### 8. Frequency Encoding
### Step 1 – Understand

Replaces each category with its frequency/proportion.

### Step 2 – Demonstrate
Apartment → 50%
Villa     → 30%
House     → 20%
### Step 3 – Implement


```python
frequency = df['Property_Type'].value_counts(
    normalize=True
)

df['Property_Type_Frequency'] = (
    df['Property_Type'].map(frequency)
)

print(df[['Property_Type', 'Property_Type_Frequency']])
```

### 9. Target Encoding
### Step 1 – Understand

Replaces each category with the average target value.

For house-price prediction:

Apartment → Average Price
Villa     → Average Price
House     → Average Price
### Step 2 – Demonstrate
Apartment → ₹80 lakhs
Villa     → ₹150 lakhs
House     → ₹60 lakhs
### Step 3 – Implement


```python
target_mean = df.groupby(
    'Property_Type'
)['Price_lakhs'].mean()

df['Property_Type_Target'] = (
    df['Property_Type'].map(target_mean)
)
```

### 10. Rare Category Grouping
### Step 1 – Understand

Combines categories with very few observations into Other.

### Step 2 – Demonstrate
Apartment → 500
Villa     → 300
Penthouse → 3
Farmhouse → 2

After grouping:

Apartment → 500
Villa     → 300
Other     → 5
### Step 3 – Implement


```python
counts = df['Property_Type'].value_counts()

rare_categories = counts[counts < 5].index

df['Property_Type_Grouped'] = (
    df['Property_Type'].replace(
        rare_categories,
        'Other'
    )
)
```

### 11. Category Cardinality
### Step 1 – Understand

### Cardinality means the number of unique categories.

## Step 2 – Demonstrate
#Property_Type → 3 categories

```python
print(df['Property_Type'].nunique())

print(df['Furnishing_Status'].nunique())

print(df['Construction_Status'].nunique())

print(df['Locality_Type'].nunique())
```

###12. Handling Unknown Categories
### Step 1 – Understand

An unknown category appears in test/production data but wasn't present during training.

### Step 2 – Demonstrate
Training:
Apartment, Villa, House

Testing:
Apartment, Villa, House, Penthouse
                              ↑
                           Unknown
### Step 3 – Implement


```python
| Technique            | Main Idea                  | When to Use          | Python                    |
| -------------------- | -------------------------- | -------------------- | ------------------------- |
| **One-Hot**          | Category → 0/1 columns     | Nominal              | `OneHotEncoder()`         |
| **Ordinal**          | Category → ordered numbers | Ordered categories   | `OrdinalEncoder()`        |
| **Label**            | Category → numeric label   | Mainly target        | `LabelEncoder()`          |
| **Frequency**        | Category → frequency       | High cardinality     | `value_counts()`          |
| **Target**           | Category → target mean     | High cardinality     | `groupby().mean()`        |
| **Rare Grouping**    | Rare → `Other`             | Rare categories      | `replace()`               |
| **Unknown Handling** | Unseen → ignored           | Test/production      | `handle_unknown='ignore'` |
| **Combining**        | Similar → one category     | Redundant categories | `replace()`               |
| **Cardinality**      | Count unique categories    | Choose encoding      | `nunique()`               |
```

## 13. Combining Categories
### Step 1 – Understand

Combining categories means merging similar categories.

### Step 2 – Demonstrate
Bungalow
Villa
Independent House

can be grouped as:

House
## Step 3 – Implement


```python
df['Property_Type_Grouped'] = df[
    'Property_Type'
].replace({
    'Bungalow': 'House',
    'Independent House': 'House'
})
```

```python
| Technique            | Main Idea                  | When to Use          | Python                    |
| -------------------- | -------------------------- | -------------------- | ------------------------- |
| **One-Hot**          | Category → 0/1 columns     | Nominal              | `OneHotEncoder()`         |
| **Ordinal**          | Category → ordered numbers | Ordered categories   | `OrdinalEncoder()`        |
| **Label**            | Category → numeric label   | Mainly target        | `LabelEncoder()`          |
| **Frequency**        | Category → frequency       | High cardinality     | `value_counts()`          |
| **Target**           | Category → target mean     | High cardinality     | `groupby().mean()`        |
| **Rare Grouping**    | Rare → `Other`             | Rare categories      | `replace()`               |
| **Unknown Handling** | Unseen → ignored           | Test/production      | `handle_unknown='ignore'` |
| **Combining**        | Similar → one category     | Redundant categories | `replace()`               |
| **Cardinality**      | Count unique categories    | Choose encoding      | `nunique()`               |
```

```python
| Column                | Type               | Recommended Encoding |
| --------------------- | ------------------ | -------------------- |
| `Property_Type`       | Nominal            | One-Hot              |
| `Furnishing_Status`   | Ordinal*           | Ordinal              |
| `Construction_Status` | Depends on meaning | One-Hot / Ordinal    |
| `Locality_Type`       | Nominal            | One-Hot              |
```
