### Date & Time Feature Engineering

1. Year

Definition: Extracts the year from a date.

Example: 2026-08-27 → 2026



```python
import pandas as pd
df = pd.read_csv("house_price_prediction_50_rows_categorical_datetime.csv")

df.columns
```

```python
print(df["Listing_Date"])
```

```python
df['Listing_Date'] = pd.to_datetime(df['Listing_Date'])
```

```python
df["Listing_Month"] = df["Listing_Date"].dt.month
df["Listing_Day"] = df["Listing_Date"].dt.day
df["Listing_DayOfWeek"] = df["Listing_Date"].dt.dayofweek
df["Listing_Quarter"] = df["Listing_Date"].dt.quarter
```

```python
## 2. Month

# Definition: Extracts the month number.

# Example: 2026-08-27 → 8

# df["Listing_Month"] = df["Listing_Date"].dt.month
```

```python
import pandas as pd

df['Listing_Date'] = pd.to_datetime(df['Listing_Date'])

df['Year'] = df['Listing_Date'].dt.year
df['Month'] = df['Listing_Date'].dt.month
df['Day'] = df['Listing_Date'].dt.day

print(df[['Listing_Date', 'Year', 'Month', 'Day']].head())
```

```python
df['Day']
```

```python
| Feature               | Short Python Code                                                              |
| --------------------- | ------------------------------------------------------------------------------ |
| **Year**              | `df['Year'] = df['Listing_Date'].dt.year`                                      |
| **Month**             | `df['Month'] = df['Listing_Date'].dt.month`                                    |
| **Day**               | `df['Day'] = df['Listing_Date'].dt.day`                                        |
| **Day of Week**       | `df['Day_of_Week'] = df['Listing_Date'].dt.dayofweek`                          |
| **Week Number**       | `df['Week_Number'] = df['Listing_Date'].dt.isocalendar().week`                 |
| **Quarter**           | `df['Quarter'] = df['Listing_Date'].dt.quarter`                                |
| **Weekend Indicator** | `df['Is_Weekend'] = (df['Listing_Date'].dt.dayofweek >= 5).astype(int)`        |
| **Month Start**       | `df['Month_Start'] = df['Listing_Date'].dt.is_month_start.astype(int)`         |
| **Month End**         | `df['Month_End'] = df['Listing_Date'].dt.is_month_end.astype(int)`             |
| **Days Since Event**  | `df['Days_Since_Event'] = (pd.Timestamp.today() - df['Listing_Date']).dt.days` |
| **Time Difference**   | `df['Time_Difference'] = df['Listing_Time'].diff().dt.total_seconds()/3600`    |
| **Seasonality**       | `df['Season'] = df['Listing_Date'].dt.month % 12 // 3`                         |
```
