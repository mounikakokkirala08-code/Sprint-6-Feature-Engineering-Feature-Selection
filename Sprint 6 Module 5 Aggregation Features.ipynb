### Aggregation Features

### Step 1 – Understand the Concept (Markdown)

Aggregation Features combine multiple values to create a meaningful summary feature.

Common aggregations:

Sum → Total value
Mean → Average value
Maximum → Highest value
Minimum → Lowest value
Count → Number of values
### Step 2 – Demonstrate the Concept (Markdown)

For the house dataset:

Bedrooms = 3
Bathrooms = 2
Parking_Spaces = 1

Total Facilities:

3 + 2 + 1 = 6

We can create:

Total_Facilities = Bedrooms + Bathrooms + Parking_Spaces
### Step 3 – Implement the Concept (Python)


```python
import pandas as pd
df = pd.read_csv("house_price_prediction_cleaned.csv")
```

```python
df['Total_Facilities'] = (df['Bedrooms']+ df['Bathrooms']+ df['Parking_Spaces'])

print(df[['Bedrooms', 'Bathrooms','Parking_Spaces', 'Total_Facilities']].head())
```

## Mean Aggregation
### Step 1 – Understand the Concept (Markdown)

Mean is the average value of a numerical feature.

Formula: Mean = Sum of Values / Number of Values
### Step 2 – Demonstrate the Concept (Markdown)

For example:

Bedrooms = 2, 3, 4

Mean = (2 + 3 + 4) / 3 = 3
### Step 3 – Implement the Concept (Python)


```python
mean_bedrooms = df['Bedrooms'].mean()

print("Average Bedrooms:", mean_bedrooms)
```

## Median Aggregation
### Step 1 – Understand the Concept (Markdown)

Median is the middle value when data is arranged in ascending order.

Formula:

Median = Middle Value
### Step 2 – Demonstrate the Concept (Markdown)

Example:

Bedrooms = 2, 3, 4

Median = 3

For an even number of values:

2, 3, 4, 6

Median = (3 + 4) / 2 = 3.5
### Step 3 – Implement the Concept (Python)


```python
median_bedrooms = df['Bedrooms'].median()

print("Median Bedrooms:", median_bedrooms)
```

## Standard Deviation
### Step 1 – Understand the Concept (Markdown)

Standard Deviation measures how much the values are spread out from the mean.

Low Standard Deviation → Values are close to the mean.
High Standard Deviation → Values are widely spread.

### Step 2 – Demonstrate the Concept (Markdown)

Example:

Bedrooms = 2, 3, 4

Mean = 3

The values are close to the mean, so the standard deviation is small.

Formula:

Standard Deviation = √(Σ(X − Mean)² / N)
### Step 3 – Implement the Concept (Python)


```python
std_bedrooms = df['Bedrooms'].std()

print("Standard Deviation:", std_bedrooms)
```

### Transaction-Level Aggregation
#### Step 1 – Understand the Concept (Markdown)

Transaction-Level Aggregation means creating summary features for each transaction using related transaction data.

Common features:

Total Amount → Sum of items
Item Count → Number of items
Average Item Price → Mean price
Maximum Item Price → Highest price
### Step 2 – Demonstrate the Concept (Markdown)

Example:

Transaction = T001
Items = 3
Prices = ₹100, ₹200, ₹300

Total Amount = ₹600
Item Count = 3
Average Price = ₹200
### Step 3 – Implement the Concept (Python)


```python
transaction_features = df.groupby('Transaction_ID').agg(
    Total_Amount=('Amount', 'sum'),
    Item_Count=('Amount', 'count'),
    Average_Amount=('Amount', 'mean'),
    Maximum_Amount=('Amount', 'max')
).reset_index()

print(transaction_features.head())
```

```python
| Aggregation Type                  | Group By         | Common Calculations | Purpose                     |
| --------------------------------- | ---------------- | ------------------- | --------------------------- |
| **Count**                         | Any column       | `count()`           | Count records/categories    |
| **Sum**                           | Numerical column | `sum()`             | Calculate total             |
| **Mean**                          | Numerical column | `mean()`            | Calculate average           |
| **Median**                        | Numerical column | `median()`          | Find middle value           |
| **Minimum**                       | Numerical column | `min()`             | Find smallest value         |
| **Maximum**                       | Numerical column | `max()`             | Find largest value          |
| **Standard Deviation**            | Numerical column | `std()`             | Measure data spread         |
| **Group-Based Aggregation**       | Category         | `groupby().agg()`   | Compare groups              |
| **Customer-Level Aggregation**    | `Customer_ID`    | Sum, Mean, Count    | Analyze customer behavior   |
| **Product-Level Aggregation**     | `Product_ID`     | Sum, Mean, Count    | Analyze product performance |
| **Transaction-Level Aggregation** | `Transaction_ID` | Sum, M_             |                             |
```
