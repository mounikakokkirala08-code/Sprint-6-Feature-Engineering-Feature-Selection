## Interaction Features

```python
Interaction Features
### Step 1 – Understand the Concept (Markdown)

Interaction Features combine two or more features to capture their combined effect on the target.

The effect of one feature may depend on another feature.

Formula: Interaction = Feature 1 × Feature 2

Example: Area_sqft × Location_Score
### Step 2 – Demonstrate the Concept (Markdown)

For the house dataset:

Area_sqft = 1500
Location_Score = 8

Interaction = 1500 × 8
            = 12000

Create:

Area_Location_Interaction = Area_sqft × Location_Score

This can help the model understand that a large house in a high-quality location may have a different effect on price than either feature alone.

### Step 3 – Implement the Concept (Python)
```

```python
import pandas as pd

df = pd.read_csv("house_price_prediction_cleaned.csv")
```

```python
df['Area_Location_Interaction'] = (df['Area_sqft'] * df['Location_Score'])

print(df[['Area_sqft','Location_Score','Area_Location_Interaction']].head()
)
```

### Feature × Feature
### Step 1 – Understand the Concept

Feature × Feature means multiplying two existing features to create a new interaction feature.

Formula: Interaction Feature = Feature 1 × Feature 2
### Step 2 – Demonstrate

For your house dataset:

Area_sqft = 1500
Bedrooms = 3

1500 × 3 = 4500

New feature:

Area_Bedroom_Interaction = Area_sqft × Bedrooms
### Step 3 – Implement


```python
df['Area_Bedroom_Interaction'] = (
    df['Area_sqft'] * df['Bedrooms']
)

print(
    df[['Area_sqft', 'Bedrooms',
        'Area_Bedroom_Interaction']].head()
)
```

## Feature + Feature
### Step 1 – Understand the Concept (Markdown)

Feature + Feature means adding two existing features to create a new feature.

Formula:

New Feature = Feature 1 + Feature 2
### Step 2 – Demonstrate the Concept (Markdown)

For your house dataset:

Bedrooms = 3
Bathrooms = 2

3 + 2 = 5

New feature:

Total_Rooms = Bedrooms + Bathrooms
### Step 3 – Implement the Concept (Python)


```python
df['Total_Rooms'] = (
    df['Bedrooms'] + df['Bathrooms']
)

print(
    df[['Bedrooms', 'Bathrooms', 'Total_Rooms']].head()
)
```

## Feature − Feature
### Step 1 – Understand the Concept (Markdown)

Feature − Feature means subtracting one feature from another to create a new feature.

Formula:

New Feature = Feature 1 − Feature 2
### Step 2 – Demonstrate the Concept (Markdown)

For your house dataset:

Bedrooms = 3
Bathrooms = 2

3 − 2 = 1

New feature:

Bedroom_Bathroom_Difference = Bedrooms − Bathrooms
### Step 3 – Implement the Concept (Python)


```python
df['Bedroom_Bathroom_Difference'] = (
    df['Bedrooms'] - df['Bathrooms']
)

print(
    df[['Bedrooms',
        'Bathrooms',
        'Bedroom_Bathroom_Difference']].head()
)
```

## Feature / Feature
### Step 1 – Understand the Concept (Markdown)

Feature / Feature means dividing one feature by another to create a ratio feature.

Formula:

New Feature = Feature 1 ÷ Feature 2
## Step 2 – Demonstrate the Concept (Markdown)

For your house dataset:

Area_sqft = 1500
Bedrooms = 3

1500 ÷ 3 = 500

New feature:

Area_per_Bedroom = Area_sqft / Bedrooms
### Step 3 – Implement the Concept (Python)


```python
df['Area_per_Bedroom'] = (
    df['Area_sqft'] / df['Bedrooms']
)

print(
    df[['Area_sqft',
        'Bedrooms',
        'Area_per_Bedroom']].head()
)
```

## Polynomial Features
### Step 1 – Understand the Concept (Markdown)

Polynomial Features create new features by using powers and combinations of existing numerical features.

They help ML models capture non-linear relationships.

Examples:

X²
X³
X₁ × X₂
### Step 2 – Demonstrate the Concept (Markdown)

For your house dataset:

Area_sqft = 1000

Polynomial features with degree 2:

Area_sqft
Area_sqft²

For two features:

Area_sqft
Bedrooms
Area_sqft²
Area_sqft × Bedrooms
Bedrooms²
### Step 3 – Implement the Concept (Python)


```python
from sklearn.preprocessing import PolynomialFeatures

poly = PolynomialFeatures(
    degree=2,
    include_bias=False
)

features = df[['Area_sqft', 'Bedrooms']]

poly_features = poly.fit_transform(features)

print(poly_features)
```

### Pairwise Interactions
## Step 1 – Understand the Concept (Markdown)

Pairwise Interactions create new features by combining every pair of selected features.

For example, with:

Area_sqft
Bedrooms
Bathrooms

Pairwise interactions include:

Area_sqft × Bedrooms
Area_sqft × Bathrooms
Bedrooms × Bathrooms
### Step 2 – Demonstrate the Concept (Markdown)

Suppose:

Area_sqft = 1500
Bedrooms = 3
Bathrooms = 2

Then:

Area × Bedrooms = 1500 × 3 = 4500

Area × Bathrooms = 1500 × 2 = 3000

Bedrooms × Bathrooms = 3 × 2 = 6

These features help the model understand combined relationships.

### Step 3 – Implement the Concept (Python)


```python
from sklearn.preprocessing import PolynomialFeatures

features = df[
    ['Area_sqft', 'Bedrooms', 'Bathrooms']
]

poly = PolynomialFeatures(
    degree=2,
    interaction_only=True,
    include_bias=False
)

interaction_features = poly.fit_transform(features)

print(poly.get_feature_names_out(
    ['Area_sqft', 'Bedrooms', 'Bathrooms']
))
```

### Domain-Based Interactions
### Step 1 – Understand the Concept (Markdown)

Domain-Based Interactions are features created by combining variables based on business or domain knowledge.

Instead of blindly combining every feature, we create interactions that make sense for the problem.

### Step 2 – Demonstrate the Concept (Markdown)

For your house-price dataset:

Area_sqft × Location_Score

A large house in a good location may have a stronger effect on price.

Other examples:

Bedrooms × Bathrooms
Area_sqft × Bedrooms
Age_years × Location_Score
Parking_Spaces × Bedrooms
### Step 3 – Implement the Concept (Python)


```python
df['Area_Location'] = (
    df['Area_sqft'] * df['Location_Score']
)

df['Bedroom_Bathroom'] = (
    df['Bedrooms'] * df['Bathrooms']
)

df['Age_Location'] = (
    df['Age_years'] * df['Location_Score']
)

df['Parking_Bedroom'] = (
    df['Parking_Spaces'] * df['Bedrooms']
)
```

```python
| Technique                     | Formula / Example                     | Purpose                                         |
| ----------------------------- | ------------------------------------- | ----------------------------------------------- |
| **Feature × Feature**         | `Area × Bedrooms`                     | Capture combined effect                         |
| **Feature + Feature**         | `Bedrooms + Bathrooms`                | Combine related features                        |
| **Feature − Feature**         | `Bedrooms − Bathrooms`                | Capture difference                              |
| **Feature / Feature**         | `Area / Bedrooms`                     | Create ratio                                    |
| **Polynomial Features**       | `Area²`, `Bedrooms²`                  | Capture nonlinear relationships                 |
| **Pairwise Interactions**     | `Area × Bedrooms`, `Area × Bathrooms` | Capture interactions between feature pairs      |
| **Domain-Based Interactions** | `Area × Location_Score`               | Create meaningful domain-specific relationships |
```
