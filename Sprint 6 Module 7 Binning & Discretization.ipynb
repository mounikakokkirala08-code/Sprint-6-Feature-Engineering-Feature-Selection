## Binning & Discretization

### What is Binning?
### Step 1 – Understand the Concept (Markdown)

Binning is a feature engineering technique that converts a continuous numerical variable into groups or ranges (bins).

Instead of using exact values, we divide values into meaningful categories.

Example:

Age
0–18    → Young
19–35   → Adult
36–60   → Middle Age
61+     → Senior
### Step 2 – Demonstrate the Concept (Markdown)

For your house dataset, we can divide Age_years into groups:

Age_years

0–5    → New
6–15   → Moderate
16–30  → Old
31+    → Very Old

So:

Age = 10 → Moderate
Age = 25 → Old
### Step 3 – Implement the Concept (Python)


```python
import pandas as pd

df = pd.read_csv("house_price_prediction_cleaned.csv")
```

```python
df['Age_Category'] = pd.cut(
    df['Age_years'],
    bins=[0, 5, 15, 30, float('inf')],
    labels=['New', 'Moderate', 'Old', 'Very Old']
)

print(df[['Age_years', 'Age_Category']].head())
```

## Equal-Width Binning
### Step 1 – Understand the Concept (Markdown)

Equal-Width Binning divides the range of values into bins having the same numerical width.

Formula:

Bin Width = (Maximum − Minimum) / Number of Bins
### Step 2 – Demonstrate the Concept (Markdown)

Suppose Age_years ranges from 0 to 40 and we want 4 bins:

0–10    → Bin 1
10–20   → Bin 2
20–30   → Bin 3
30–40   → Bin 4

Each bin has the same width: 10 years.

### Step 3 – Implement the Concept (Python)

Using pd.cut():


```python

df['Age_Bin'] = pd.cut(
    df['Age_years'],
    bins=4
)

print(df[['Age_years', 'Age_Bin']].head())
```

### Equal-Frequency Binning
### Step 1 – Understand the Concept (Markdown)

Equal-Frequency Binning divides data into bins so that each bin contains approximately the same number of observations.

It is also called Quantile Binning.

Example:

100 observations → 4 bins

Bin 1 → 25 observations
Bin 2 → 25 observations
Bin 3 → 25 observations
Bin 4 → 25 observations
### Step 2 – Demonstrate the Concept (Markdown)

Suppose Price_lakhs has 100 houses.

Using 4 bins:

Q1 → Lowest 25% prices
Q2 → 25–50%
Q3 → 50–75%
Q4 → Highest 25%

The number of houses in each bin is approximately equal, but the value ranges may be different.

### Step 3 – Implement the Concept (Python)



```python


df['Price_Bin'] = pd.qcut(
    df['Price_lakhs'],
    q=4,
    labels=['Low', 'Medium', 'High', 'Very High']
)

print(
    df[['Price_lakhs', 'Price_Bin']].head()
)
```

Domain-Based Binning
### Step 1 – Understand the Concept

Definition:
Domain-Based Binning means grouping continuous numerical values into meaningful categories based on domain knowledge.

Example:

Age → Age Group

Instead of using exact ages, create:

18–30  → Young
31–50  → Middle
51+    → Senior
### Step 2 – Demonstrate the Concept

For your House Price dataset, Age_years can be grouped:

Age_years	Property Age Group
0–5	New
6–10	Recent
11–20	Old
21+	Very Old

This is useful when these ranges have real-world meaning.

### Step 3 – Implement the Concept


```python
bins = [0, 5, 10, 20, float("inf")]
labels = ["New", "Recent", "Old", "Very Old"]

df["Property_Age_Group"] = pd.cut(
    df["Age_years"],
    bins=bins,
    labels=labels
)
```
