## Character Count
### Step 1 – Understand the Concept

Definition:
Character Count is the number of characters present in a text string.

It can include letters, numbers, spaces, and punctuation depending on how you calculate it.

### Step 2 – Demonstrate the Concept

Example:

Text = "Beautiful House"

Character count including the space:

15

Without spaces:

14
### Step 3 – Implement the Concept


```python
df["Character_Count"] = df["Text"].str.len()

#Without spaces:

df["Character_Count"] = df["Text"].str.replace(" ", "").str.len()
```

## Word Count
### Step 1 – Understand the Concept

Definition:
Word Count is the number of words present in a text string.

### Step 2 – Demonstrate the Concept

Example:

Text = "This house is beautiful"

Word count:

This → 1
house → 2
is → 3
beautiful → 4

Word Count = 4

### Step 3 – Implement the Concept


```python
df["Word_Count"] = df["Text"].str.split().str.len()
```

## Number of Special Characters
### Step 1 – Understand the Concept

Definition:
Number of Special Characters is the count of characters that are not letters, numbers, or spaces.

Examples: @, #, !, $, %, &

### Step 2 – Demonstrate the Concept

Example:

"Hello! House @ ₹50"

Special characters include:

!  @  ₹

Count = 3

### Step 3 – Implement the Concept


```python
df["Special_Char_Count"] = df["Text"].str.count(r"[^a-zA-Z0-9\s]")
```

## Bag of Words (BoW)
### Step 1 – Understand the Concept

Definition:
Bag of Words converts text into numbers based on word frequency.

It ignores grammar and word order and focuses mainly on which words appear and how often.

### Step 2 – Demonstrate the Concept

Texts:

"This house is good"
"This house is big"

Vocabulary:

[this, house, is, good, big]

Representation:

Text	this	house	is	good	big
Text 1	1	1	1	1	0
Text 2	1	1	1	0	1
### Step 3 – Implement the Concept


Simple meaning:

Bag of Words = Convert text into numerical word-count vectors.

TF-IDF
### Step 1 – Understand the Concept

Definition:
TF-IDF (Term Frequency–Inverse Document Frequency) converts text into numerical values based on how important a word is in a document.

It gives higher importance to words that are frequent in one document but rare across many documents.

### Step 2 – Demonstrate the Concept

Example:

Document 1: "House has garden"
Document 2: "House has pool"

House appears in both documents → less important.

Garden appears only in Document 1 → more important.

Common word → Lower TF-IDF
Rare/important word → Higher TF-IDF
### Step 3 – Implement the Concept


```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer()

X = vectorizer.fit_transform(df["Text"])
```

```python
N-Grams
### Step 1 – Understand the Concept

Definition:
N-Grams are groups of N consecutive words taken from a text.

Unigram → 1 word
Bigram → 2 words
Trigram → 3 words
### Step 2 – Demonstrate the Concept

Text:

"This house is very beautiful"

Unigrams:

This | house | is | very | beautiful

Bigrams:

This house
house is
is very
very beautiful

Trigrams:

This house is
house is very
is very beautiful
### Step 3 – Implement the Concept
from sklearn.feature_extraction.text import CountVectorizer

vectorizer = CountVectorizer(ngram_range=(1, 2))

X = vectorizer.fit_transform(df["Text"])
```
